```python
import random

def extract_relations_and_save(input_file, output_file, num_nodes=10, level=2):
    # Read data from file
    with open(input_file, 'r') as file:
        lines = file.read().strip().split('\n')
    
    # Shuffle lines to ensure randomness
    random.shuffle(lines)

    # Select a random sample of lines
    sample = random.sample(lines, min(num_nodes, len(lines)))

    # Extract initial nodes from the sample
    initial_nodes = {line.split('\t')[0] for line in sample}

    # Initialize the set of nodes to include initial nodes
    nodes = set(initial_nodes)
    all_related_lines = set(sample)  # Include initial lines

    # Expand the search to include relations at the specified level
    for _ in range(level):
        current_level_nodes = set()
        # Find all lines that involve any of the nodes from the last iteration
        for line in lines:
            node_id = line.split('\t')[0]
            if node_id in nodes:
                all_related_lines.add(line)
                # Add new nodes from this line
                related_node = line.split('\t')[-1]
                current_level_nodes.add(related_node)
        # Update nodes with new nodes found at this level
        nodes.update(current_level_nodes)

    # Write all related lines to the output file
    with open(output_file, 'w') as f:
        for line in all_related_lines:
            f.write(line + '\n')

# Example usage:
extract_relations_and_save(input_file='/Users/abhayvaghasiya/Desktop/WORK/Datasets/Datasets_Perturbed/FB15k-237/0.0/train.txt', 
                           output_file='/Users/abhayvaghasiya/Desktop/WORK/Datasets/Fb15k-237_train.txt', 
                           num_nodes=10, level=2)
```

```python
import random

def split_data(input_file, train_file, validation_file, test_file, validation_ratio=0.1, test_ratio=0.1):
    # Read the data from the input file
    with open(input_file, 'r') as file:
        lines = file.readlines()
    
    # Shuffle the lines to ensure random distribution
    random.shuffle(lines)
    
    # Calculate the number of lines for validation and testing
    total_lines = len(lines)
    validation_size = int(total_lines * validation_ratio)
    test_size = int(total_lines * test_ratio)
    
    # Split the data
    validation_data = lines[:validation_size]
    test_data = lines[validation_size:validation_size + test_size]
    training_data = lines[validation_size + test_size:]
    
    # Write training data to the training file
    with open(train_file, 'w') as t_file:
        t_file.writelines(training_data)
    
    # Write validation data to the validation file
    with open(validation_file, 'w') as v_file:
        v_file.writelines(validation_data)
    
    # Write test data to the test file
    with open(test_file, 'w') as tst_file:
        tst_file.writelines(test_data)

# Example usage:
split_data(input_file='/Users/abhayvaghasiya/Desktop/WORK/Datasets/train.txt',
           train_file='/Users/abhayvaghasiya/Desktop/WORK/Datasets/new_train.txt',
           validation_file='/Users/abhayvaghasiya/Desktop/WORK/Datasets/valid.txt',
           test_file='/Users/abhayvaghasiya/Desktop/WORK/Datasets/test.txt',
           validation_ratio=0.1,
           test_ratio=0.1)
```
